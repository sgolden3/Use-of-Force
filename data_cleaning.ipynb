```python
# Remove leading/trailing spaces

# Standardize text (for example: convert names or categories to consistent casing)
```

```python
# Strip spaces from ALL text columns
obj_cols = clean.select_dtypes(include=["object"]).columns.tolist()
for c in obj_cols:
    clean[c] = clean[c].astype(str).str.strip()
    
clean.head()
```

```python
# Convert date columns to datetime objects (Only showing as an example)
if "Occurred_date_time" in clean.columns:
    clean["Occurred_date_time"] = pd.to_datetime(clean["Occurred_date_time"], errors="coerce")

clean[["Occurred_date_time"]].head() if "Occurred_date_time" in clean.columns else clean.head()
```

```python
clean.info()
```

```python
# Rename Occurred_date_time to Date_Time
```

```python
clean = clean.rename(columns={"Occurred_date_time": "Date_Time"})
clean.info()
```

```python
clean.head()
```

```python
#Added 
if "Date_Time" in clean.columns:
    clean["Year"] = clean["Date_Time"].dt.year
    clean["Month"] = clean["Date_Time"].dt.month
clean.head()
```

```python
clean.to_csv("cleaned_data.csv", index=False)
print("Saved: cleaned_data.csv")
clean.head()
```
